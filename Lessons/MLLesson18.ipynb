```python
def Inv_boladimi(x):
    rows, columns=x.shape
    if rows!=columns:
        print('Bu kvadrat funksiya emas')
    elif np.linalg.det(x)==0:
        print('Bu matritsa determinanti 0 ga teng')
    else:
        return np.linalg.inv(x)
        
```
